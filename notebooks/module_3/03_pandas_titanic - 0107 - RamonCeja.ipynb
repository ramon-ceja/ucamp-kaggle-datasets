```python
import pandas as pd
```

```python
# Crear DataFrame con los datos proporcionados
dataTitanic = pd.read_csv("../../data/titanic.csv", index_col=0)

# Validamos
print(dataTitanic.head())
```

```python
#Muestra las dimensiones del DataFrame.
print(f"Las dimensiones de nuestro DataFrame son: {dataTitanic.shape}")
```

```python
#Muestra los datos que contiene.
print(f"Los datos que contiene nuestro DataFrame son:\n{dataTitanic.dtypes}")
```

```python
#Muestra los nombres de sus columnas y filas.
print(f"Las columnas son:\n{dataTitanic.columns}\n y las filas son:\n{dataTitanic.index}")
```

```python
#Muestra las primeras 10 filas y las 10 últimas filas.
print(f"Las primeras 10 filas del DataFrame son:\n{dataTitanic.head(10)}")
print(f"Las ultimas 10 filas del DataFrame son:\n{dataTitanic.tail(10)}")

```

```python
#Muestra las filas pares del DataFrame.
print(f"Las filas pares del DataFrame son:\n{dataTitanic.iloc[dataTitanic.index % 2 == 0]}")

```

```python
#Muestra los nombres de las personas que iban en segunda clase ordenadas alfabéticamente.
# dolor de cabeza
print(f"Nombres de las personas en Segunda Clase:\n{dataTitanic.loc[dataTitanic["Pclass"] == 2, "Name"].sort_values()}")

```

```python
#Muestra el porcentaje de personas que sobrevivieron y murieron.
allPassengers = len(dataTitanic)
passengerLives = len(dataTitanic[dataTitanic["Survived"] == 1])
passengerDeath = len(dataTitanic[dataTitanic["Survived"] == 0])

passengerLives = (passengerLives / allPassengers) * 100
passengerDeath = (passengerDeath / allPassengers) * 100

print(f"Sobrevivieron: {passengerLives:.2f}%")
print(f"Murieron: {passengerDeath:.2f}%")
print(f"Total: {passengerDeath + passengerLives}")
```

```python

#Muestra el porcentaje de personas que sobrevivieron en cada clase.
passengersFirstClass = len(dataTitanic[dataTitanic["Pclass"] == 1])
passengersSecondClass = len(dataTitanic[dataTitanic["Pclass"] == 2])
passengersTreeClass = len(dataTitanic[dataTitanic["Pclass"] == 3])

survivedFirstClass = len(dataTitanic[(dataTitanic["Pclass"] == 1) & (dataTitanic["Survived"] == 1)])
survivedSecondClass = len(dataTitanic[(dataTitanic["Pclass"] == 2) & (dataTitanic["Survived"] == 1)])
survivedTreeClass = len(dataTitanic[(dataTitanic["Pclass"] == 3) & (dataTitanic["Survived"] == 1)])

survivedFirstClass = (survivedFirstClass / passengersFirstClass) * 100
survivedSecondClass = (survivedSecondClass / passengersSecondClass) * 100
survivedTreeClass = (survivedTreeClass / passengersTreeClass) * 100


print(f"Porcentaje de supervivencia en Primera Clase: {survivedFirstClass:.2f}%")
print(f"Porcentaje de supervivencia en Segunda Clase: {survivedSecondClass:.2f}%")
print(f"Porcentaje de supervivencia en Tercera Clase: {survivedTreeClass:.2f}%")


```

```python
#Elimina del DataFrame los pasajeros con edad desconocida.
# No supe si teniamos que continuar los demas apartados con el DATA nuevo o el original, pero opte por avanzar con el nuevo
newData = dataTitanic.dropna(subset=["Age"])
passengerNotAge = len(dataTitanic) - len(newData)

print(f"Filas originales: {len(dataTitanic)}")
print(f"Filas sin edad desconocida: {len(newData)}")
print(f"Diferencia: {passengerNotAge}")

```

```python
#Muestra la edad media de las mujeres que viajaban en cada clase.
allFemales = len(newData[newData["Sex"] == "female"])
allFemalesData = (newData[newData["Sex"] == "female"])

femalesFirstClass = (allFemalesData[allFemalesData["Pclass"] == 1])
femalesSecondClass = (allFemalesData[allFemalesData["Pclass"] == 2])
femalesTreeClass = (allFemalesData[allFemalesData["Pclass"] == 3])

print(f"La media de edad en mujeres de la Primera Clase es: {femalesFirstClass['Age'].mean():.2f}")
print(f"La media de edad en mujeres de la Segunda Clase es: {femalesSecondClass['Age'].mean():.2f}")
print(f"La media de edad en mujeres de la Tercera Clase es: {femalesTreeClass['Age'].mean():.2f}")


```

```python
#Muestra por pantalla el porcentaje de menores y mayores de edad que sobrevivieron en cada clase.
# se considara mayor de edad a los 18 años.
kidsAll = (newData[newData["Age"] <= 17])
adultsAll = (newData[newData["Age"] >= 18])

kidsInFirst = len(kidsAll[kidsAll["Pclass"] == 1])
kidsInSecond = len(kidsAll[kidsAll["Pclass"] == 2])
kidsInTree = len(kidsAll[kidsAll["Pclass"] == 3])

adultsInFirst = len(adultsAll[adultsAll["Pclass"] == 1])
adultsInSecond = len(adultsAll[adultsAll["Pclass"] == 2])
adultsInTree = len(adultsAll[adultsAll["Pclass"] == 3])

kidsLive = (kidsAll[kidsAll["Survived"] == 1])
adultsLive = (adultsAll[adultsAll["Survived"] == 1])

kidsFirstClass = len(kidsLive[kidsLive["Pclass"] == 1])
kidsSecondClass = len(kidsLive[kidsLive["Pclass"] == 2])
kidsTreeClass = len(kidsLive[kidsLive["Pclass"] == 3])

adultsFirstClass = len(adultsLive[adultsLive["Pclass"] == 1])
adultSecondClass = len(adultsLive[adultsLive["Pclass"] == 2])
adultTreeClass = len(adultsLive[adultsLive["Pclass"] == 3])

porcKidsFirst = (kidsFirstClass / kidsInFirst) * 100
porcKidsSecond = (kidsSecondClass / kidsInSecond) * 100
porcKidsTree = (kidsTreeClass / kidsInTree) * 100

porcAdultsFirst = (adultsFirstClass / adultsInFirst) * 100
porcAdultSecond = (adultSecondClass / adultsInSecond) * 100
porcAdultTree = (adultTreeClass / adultsInTree) * 100

print(f"En Primera Clase sobrevivieron:\n El {porcKidsFirst:.2f}% de los niños y el {porcAdultsFirst:.2f}% de los adultos\n")

print(f"En Segunda Clase sobrevivieron:\n El {porcKidsSecond:.2f}% de los niños y el {porcAdultSecond:.2f}% de los adultos\n")

print(f"En Tercera Clase sobrevivieron:\n El {porcKidsTree:.2f}% de los niños y el {porcAdultTree:.2f}% de los adultos\n")
```
