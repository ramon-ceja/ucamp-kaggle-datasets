```python
import numpy as numpy
```

```python

# cargar el archivo de datos
arr1 = numpy.genfromtxt("../../data/reto_numpy.txt")
```

```python
# Crear un DataFrame a partir de un arreglo numpy
import pandas as pd

dic ={"A": arr1[:,0],
     "B": arr1[:,1],
     "C": arr1[:,2]}

df = pd.DataFrame(dic)

df.head()
```

```python
#2 Estadísticas de la columna A y metodos
#Suma
a = df.A.sum()
print(a)

#Mínimo
b =df.A.min()
print(b)

#Máximo
c = df.A.max()
print(c)

d = df.A.cumsum()
print(d)

e = df.A.mean()
print(e)

#Mediana
f = df.A.median()
print(f)

#Desviación Estandar
g = df.A.std()
print(g)
```
