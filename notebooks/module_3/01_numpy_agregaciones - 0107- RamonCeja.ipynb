```python
# importa la librería de Numpy

import numpy as np
```

```python
# cargar el archivo de datos

arr1 = np.genfromtxt("../../data/reto_numpy.txt")
```

```python
# Imprime la clase de arr1 y las dimensiones

print(type(arr1))
print(arr1.ndim)
```

```python
# Operaciones sobre la primera columna de arr1

a = arr1[:,0].sum()
print(a)

b = arr1[:,0].min()
print(b)

c = arr1[:,0].max()
print(c)

d = arr1[:,0].cumsum()
print(d)

e = arr1[:,0].mean()
print(e)
#
f = np.median(arr1[:,0])
print(f)

g = arr1[:,0].std()
print(g)
```
