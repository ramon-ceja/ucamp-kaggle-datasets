# Modulo 4.2: Sistema de Cajero Automático (ATM)

Simulación interactiva basada en POO que interconecta la clase `Client` (manejo de balances individuales) con la clase `ATM` (gestión de sesiones múltiples, menú de operaciones y cálculo de flujos totales al cierre del día).s.


En un cajero hay clientes que pueden hacer depósitos y retiros de dinero. Además al final del día, se calcula  la cantidad de dinero que se ha depositado.

Se deberán crear dos clases: La clase cliente y la clase cajero.

La clase cliente tendrá los atributos nombre, cantidad y los métodos __init__, depositar, retirar, mostrar_todo.

La clase  cajero tendrá como atributos 3 objetos de la clase cliente y los métodos __init__, operar y deposito_total.

```python
class Client:

    def __init__(self, name):
        if name:
            self.balance = 0
            self.name = name
            print(f"Creaste tu cuenta con éxito {self.name}")
        else:
            print("No deben estar vacíos tus datos")

    def deposit(self, deposit=""):
        if deposit and deposit > 0:
            self.balance += deposit
            print(f"Saldo Actualizado {self.name}, tienes: ${self.balance}")
        else:
            print("No puedes realizar ingresos menores de $ 0.00")

    def withdraw(self, withdraw=""):
        if withdraw and withdraw > 0:
            if self.balance >= withdraw:
                self.balance -= withdraw
                print(f"Saldo retirado exitosamente\n {self.name} tu nuevo saldo es: {self.balance}")
                return withdraw
            else:
                print(f"Losiento {self.name}, cuentas con fondos suficientes")
                return 0
        else:
            print("No puedes retirar un monto menor a $ 0.00")
            return 0

    def showMyBalance(self):
        print(f"Hola {self.name} tu saldo actual es de: ${self.balance}")


class ATM:
    def __init__(self, client1, client2, client3):
        self.client1 = client1
        self.client2 = client2
        self.client3 = client3
        self.clients = []
        #print(self.client1.name)
        self.totalWithdrawn = 0

    def operation(self):
        menuClients = {
            "1": self.client1,
            "2": self.client2,
            "3": self.client3
        }

        while True:
            print("\n¿Quien va a realizar la operación?")
            print(f"1- {self.client1.name}")
            print(f"2- {self.client2.name}")
            print(f"3- {self.client3.name}")
            print("4- Salir")

            opcionCliente = input("Selecciona una opcion: ")

            if opcionCliente == "4":
                print("Finalizando operaciones del dia")
                break

            if opcionCliente in menuClients:
                activeClient = menuClients[opcionCliente]
            else:
                print("Opcion no valida.")
                continue
            print(f"\n[Sesión: {activeClient.name}] ¿Que deseas hacer?")
            print("1- Depositar")
            print("2- Retirar")
            print("3- Ver Saldo")
            opcionAccion = input("Selecciona una opcion: ")

            if opcionAccion == "1":
                monto = float(input("Monto a depositar: $"))
                activeClient.deposit(monto)

            elif opcionAccion == "2":
                monto = float(input("Monto a retirar: $"))
                amountDraw = activeClient.withdraw(monto)
                self.totalWithdrawn += amountDraw

            elif opcionAccion == "3":
                activeClient.showMyBalance()
            else:
                print("Operacion invalida")

    def depositTotal(self):
        totalInAccounts = self.client1.balance + self.client2.balance + self.client3.balance
        totalDepositedFlow = totalInAccounts + self.totalWithdrawn
        print(f"Saldo retirado por clientes:  ${self.totalWithdrawn}")
        print(f"Saldo actual en cuentas: ${totalInAccounts}")
        print(f"Total bruto depositado: ${totalDepositedFlow}")
```

```python
client1 = Client("Ray")
client2 = Client("Manuel")
client3 = Client("Fernanda")
```

```python
client1.deposit(50)
client2.deposit(1000)
client2.deposit(500)
client3.deposit(2500)
client1.deposit(150)
client3.deposit(500)
```

```python
client1.withdraw(50)
client2.withdraw(500)
client3.withdraw(2500)
client3.withdraw(50)
client2.withdraw(500)
```

```python
client1.showMyBalance()
client2.showMyBalance()
client3.showMyBalance()
```

```python
newATM = ATM(client1, client2, client3)
```
