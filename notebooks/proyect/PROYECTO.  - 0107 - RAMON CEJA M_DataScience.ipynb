# 3. Explorando el analisis de datos

**3.1 Tablas de las diversas categorias**

````python

# Este fragmento de còdgio es una adaptacion recuperado del ejemplo que proporcional Kaggle EDA
# inicializamos el dashboard con una cuadricula de 5 filas y 2 columnas
figDashboard = make_subplots(
    rows=5, cols=2,
    subplot_titles=[f"Distribution of {col}" for col in catCols],
    specs=[[{"type": "domain"}, {"type": "xy"}],
           [{"type": "xy"},     {"type": "xy"}],
           [{"type": "xy"},     {"type": "xy"}],
           [{"type": "xy"},     {"type": "xy"}],
           [{"type": "xy"},     {"type": "xy"}]]
)

# agregamos una grafica de pastel vacia en la primera posiciòn como marcador de posicion
figDashboard.add_trace(
    go.Pie(
        labels=[],
        values=[],
        name="",
        title={
            "text": "Count plot for<br>Categorical variables",
            "font": {"size": 20, "family": "Arial Black"}
        }
    ),
    row=1, col=1
)

# removemos sel titulo automatico del primer subplot para usar el titulo personalizado del pie chart
figDashboard.layout.annotations[0].text = ""

# definimos las posiciones exactas de fila y columna para cada grafica en el dashboard
gridPositions = [(0,1,2), (1,2,1), (2,2,2), (3,3,1), (4,3,2), (5,4,1), (6,4,2), (7,5,1), (8,5,2)]

# iteramos sobre las posiciones para construir y agregar cada grafica de barras
for idx, targetRow, targetCol in gridPositions:
    colName = catCols[idx]
    dfTemp = datos[[colName]].copy()

    # acortamos el texto de machine learning para que no sature el eje x de la grafica
    if idx == 3:
        dfTemp[colName] = dfTemp[colName].astype(str).str.replace('Machine Learning', 'ML')

    # calculamos la frecuencia de cada categoria en la columna
    valueCounts = dfTemp[colName].value_counts()

    # filtramos para mostrar solo el top 9 de categorias en las columnas mas extensas
    if idx in [3, 5, 7]:
        valueCounts = valueCounts.head(9)

    xValues = valueCounts.index.astype(str).tolist()
    yValues = valueCounts.values.tolist()

    # añadimos la grafica de barras al subplot correspondiente con un mapa de color degradado
    figDashboard.add_trace(
        go.Bar(
            x=xValues,
            y=yValues,
            marker={
                "color": yValues,
                "colorscale": "Sunset_r",
                "showscale": False
            },
            showlegend=False
        ),
        row=targetRow,
        col=targetCol
    )

# configuramos el tamaño total del lienzo y aplicamos un tema claro limpio
figDashboard.update_layout(
    height=1600,
    width=1100,
    template="plotly_white",
    showlegend=False
)

# inclinamos las etiquetas del eje x en una grafica especifica para mejorar la lectura del texto
figDashboard.update_xaxes(row=3, col=1, tickangle=15)

# desplegamos el dashboard completo con todos sus subplots
#figDashboard.write_html("dashboard.html", auto_open=True)
#IFrame("dashboard.html", width=500, height=500)
figDashboard.show()
````

**3.2 Analisis temporal del salario en Dolares**

````python
# convertimos el año a texto string para que plotly lo trate como variable categorica en el eje x
datos['yearStr'] = datos['work_year'].astype(str)

# creamos el histograma grafico de barras agregadas utilizando plotly express
estadistica1 = px.histogram(
    datos,
    x="yearStr",
    y="salary_in_usd",
    nbins=30,
    color="yearStr",
    histfunc="avg",
    title="Salarios por año en Ciencia de Datos",
    labels={"salary_in_usd": "Salario (USD)", "yearStr": "año"},
    color_discrete_sequence=['#49e3d0', '#AB1B98', '#BFBF2A']
)

# personalizamos el diseño layout del grafico interactivo
estadistica1.update_layout(
    title_x=0.5,
    xaxis_title="Año Laboral",
    yaxis_title="Salario Promedio (USD)",
    xaxis={
        "tickmode": "linear",
        "tick0": 2020,
        "dtick": 1
    },
    template="plotly_dark"
)

# desplegamos el grafico
#estadistica1.write_html("grafico.html", include_plotlyjs="cdn", auto_open=True)
#iframe = IFrame(src="grafico.html", width=500, height=600)
estadistica1.show()
````

>>> **insight de la gràfica de salarios promedio por año** Demostramos como fue creciendo el salario en los trabajos de ciencias de datos a lo largo de los años 2020, 2021 y 2022

3.3 Paises con el salario más alto


````python
# agrupamos los datos por pais de la empresa para calcular el salario promedio en dolares
salarioPais = datos.groupby(by='company_location')['salary_in_usd'].mean().reset_index().sort_values(by='salary_in_usd', ascending=False)

# escalamos los valores dividiendo entre mil para expresar los salarios en k de dolares
salarioPais['salary_in_usd'] = salarioPais['salary_in_usd']/1000

# filtramos para conservar solo los paises cuyo salario promedio sea mayor a 80k
condicion = salarioPais ['salary_in_usd'] > 80.00
salarioPais = salarioPais[condicion]
````

````python
# creamos la grafica de barras horizontales utilizando plotly express para ver los salarios por pais
estadistica2 = px.bar(
    salarioPais,
    y="company_location",
    x="salary_in_usd",
    orientation="h",
    color="salary_in_usd",
    color_continuous_scale="Sunsetdark",
    title="Salario promedio por pais que ofrecen empleao (Más altos)",
    labels={"salary_in_usd":"Salario Promedio (miles de USD)", "company_location":"Pais"},
    text_auto="2.f",
)

# configuramos el diseño optimizando el orden de las barras y aplicandio el tema oscuro
estadistica2.update_layout(
    title_x=0.5,
    xaxis_title="Salario Promedio (miles de USD)",
    yaxis_title="Pais",
    template="plotly_dark",
    yaxis={'categoryorder':'total ascending'},
)

# desplegamos el grafico
# estadistica2.write_html("grafico2.html", include_plotlyjs="cdn", auto_open=True)
#iframe2 = IFrame(src="grafico2.html", width=500, height=600)
estadistica2.show()
````

>>> **insight de salarios por pais:** al filtrar los paises con salarios promedio mas altos destaca que las empresas ubicadas en rusia (RU), estados unidos (US) y nueva zelanda (NZ) lideran el mercado ofreciendo las mejores compensaciones economicas en el area de datos

# 4. Estimación de densidad de distribución salarial

````python

# Eliminamos los valores nulos de la columna de salarios
SalarioDatos = datos['salary_in_usd'].dropna()

# Creamos la gráfica de distribución moderna combinando histograma y caja marginal
estadisticaKde = px.histogram(
    datos.dropna(subset=['salary_in_usd']),
    x="salary_in_usd",
    nbins=80,
    marginal="box",
    title="4.1 Estimación de Densidad y Distribución de Salarios Globales",
    color_discrete_sequence=["#f39c12"]
)

# Configuramos el diseño del gráfico en modo oscuro
estadisticaKde.update_layout(
    title_x=0.5,
    template="plotly_dark",
    xaxis_title="Salario Anual (USD)",
    yaxis_title="Frecuencia",
)

# Exportamos, abrimos en el navegador e incrustamos el IFrame
#estadisticaKde.write_html("graficoKde.html", include_plotlyjs="cdn", auto_open=True)
#display(IFrame(src="graficoKde.html", width=700, height=600))
estadisticaKde.show()
````

>>> **insight de la curva de densidad kde:** la grafica de estimacion de densidad confirma de manera visual que los salarios globales no se distribuyen de forma simetrica sino que presentan un sesgo a la derecha con un pico muy marcado entre los 50 mil y 130 mil dolares anuales lo que demuestra cual es el rango de sueldo mas comun en la industria

**4.1 Correlaciòn de variables de paga por año**

````python
columnas_numericas = datos.select_dtypes(include=['int64', 'float64'])
matriz_pearson = columnas_numericas.corr(method='pearson')
matriz_pearson
````

>>> **insight de correlacion de variables:** al calcular la matriz de pearson se observa que los coeficientes entre las variables cuantitativas son muy cercanos a cero esto indica que no existe una relacion lineal fuerte entre el año el porcentaje de trabajo remoto y el salario en dolares por lo que el sueldo depende mas de factores cualitativos como el puesto y la experiencia del empleado
