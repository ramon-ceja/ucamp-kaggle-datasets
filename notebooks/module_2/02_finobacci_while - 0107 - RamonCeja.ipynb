## Reto Número 2.2 : Sucesión de Fibonacci con Ciclo While

Variante de la función Fibonacci utilizando un bucle `while` que continúa ejecutándose de manera dinámica hasta que el valor generado supera un límite establecido (en este caso, mayor a 150).

```python
def fibonacci(a):
  respuesta = 0
  lista = [0]
  while respuesta <= a:
    if respuesta == 0:
      respuesta = lista[-1] + 1
    else:
      respuesta = lista[-1] + lista[-2]

    lista.append(respuesta)
    print(respuesta)

  return respuesta
```

```python
fibonacci(150)
```
