## Reto Número 2.3 : Granos Ajedrez

Para cada cuadro del tablero de ajedrez, cuántos granos de trigo se usan cuando se duplica: 1 grano para el primer cuadro, 2 granos para el segundo cuadro, 4 granos para el tercer cuadro, etc. Realiza el calculo correspondiente.

```python
# granos por cuadro en tablero de ajedrez

def tablero(cuadros):
  a = 0
  b = 0
  print(f"El tablero a generar es de: {cuadros} cuadros\n")
  for i in range(1,cuadros+1):
    if i == 1:
      a = i
    elif i == 2:
      a = i
    elif i!=1 and i!=2:
      a = a*2
    b+=a
    print(f"En el cuadro número {i} hay: {a} granos")
```

```python
tablero(24)
```
